### Segment Logic

| Risk Level      | Credit Score Range |
| --------------- | -----------------: |
| **High Risk**   |            `< 580` |
| **Medium Risk** |        `580 – 699` |
| **Low Risk**    |            `≥ 700` |

### Create Risk Segmentation Table

```python
risk_segments_query = """
DROP TABLE IF EXISTS dbo.dw_risk_segments;

SELECT
    Customer_ID,
    Loan_ID,
    Credit_Score_Final,
    CASE 
        WHEN Credit_Score_Final < 580 THEN 'High Risk'
        WHEN Credit_Score_Final BETWEEN 580 AND 699 THEN 'Medium Risk'
        ELSE 'Low Risk'
    END AS Risk_Level,
    Loan_Amount,
    Interest_Rate,
    Annual_Income,
    Loan_Term_Months,
    Is_Default,
    Total_Payments,
    Avg_Payment_Amount,
    Late_Payment_Rate
INTO dbo.dw_risk_segments
FROM dbo.dw_loan_analysis;
"""
```

```python
with engine.begin() as conn:
    conn.execute(text(risk_segments_query))
```

```python
risk_segments_df = pd.read_sql("SELECT TOP 10 * FROM dbo.dw_risk_segments;", engine)
```
